```python
import glob
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
#import tifffile
import cv2
from PIL import Image

from skimage.color import label2rgb
import rasterio
from rasterio.enums import ColorInterp

import imageio.v3 as imageio

from tqdm import tqdm
```

```python
image_filename = 'calibration_data/0399_cloudless.tif'
image_rgb_png_filename = image_filename.replace('tif', 'png')
image = imageio.imread(image_filename)[:, :, :3]
im = Image.fromarray(image.astype(np.uint8), 'RGB')
im.save(image_rgb_png_filename)
```

```python
data = pd.read_csv('calibration_data/calibration_points_with_sampled_coords.csv')
```

```python
image_filenames = data["image_filename"]

for i in tqdm(range(len(image_filenames))):
    image_filename = image_filenames[i]
    image_rgb_png_filename = image_filename.replace('tif', 'png').replace('images', 'images_rgb_png')
    image = imageio.imread(image_filename)[:, :, :3]
    im = Image.fromarray(image.astype(np.uint8), 'RGB')
    im.save(image_rgb_png_filename)
```
