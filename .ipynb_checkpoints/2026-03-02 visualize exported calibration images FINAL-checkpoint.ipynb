```python
import glob
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
#import tifffile
import cv2

from skimage.color import label2rgb
import rasterio
from rasterio.enums import ColorInterp

import imageio.v3 as imageio
```

```python
image_filenames = glob.glob(os.path.join('calibration_data/images', "*.tif"))
```

```python
len(image_filenames)
```

```python
image_with_sampled_point_filenames = glob.glob(os.path.join('calibration_data/images_with_sampled_point', "*.tif"))
```

```python
len(image_with_sampled_point_filenames)
```

```python
%%capture --no-stdout --no-display

import random
np.random.seed(0)
random.seed(0)

#image_filenames = random.sample(image_filenames, 10)

for image_filename in image_filenames:
    #metadata = imageio.immeta(image_filename)
    #print(metadata)
    
    image = imageio.imread(image_filename)
    rgb_img = image[:, :, :3]
    plt.imshow(rgb_img)
    plt.axis('off')
    plt.show()
        
    image_with_sampled_point_filename = image_filename.replace("images", "images_with_sampled_point")
    image = imageio.imread(image_with_sampled_point_filename)
    rgb_img = image[:, :, :3]
    plt.imshow(rgb_img)
    plt.axis('off')
    plt.show()

    sampled_coord = np.argwhere(image[:, :, 0]==300)[0]
    print(sampled_coord)

    break
```
