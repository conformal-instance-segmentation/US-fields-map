```python
import numpy as np
np.bool = np.bool_ 
import pandas as pd
import os
from tqdm import tqdm

from sklearn.metrics import matthews_corrcoef

import random

import rasterio
from rasterio.enums import ColorInterp
from rasterio.features import shapes
from scipy import ndimage

import matplotlib.pyplot as plt
%matplotlib inline

import higra as hg
import torch
import pickle

import itertools

import imageio.v3 as imageio
from skimage.color import label2rgb
import cv2
from skimage.segmentation import mark_boundaries
from skimage.segmentation import find_boundaries

from shapely.geometry import shape
import shapely
from shapely import Polygon, orient_polygons
from shapely.wkt import loads
import geopandas as gpd
import pyproj

import ast

import glob

import multiprocessing
```

```python
filenames = []
for lat in range(23, 52):
    for lon in range(64, 127):
        filename = f"baseline_comparison_data/taylor_geospatial/all_fields_2024/2024_N{lat}W{lon:03d}.parquet"
        filenames.append(filename)
```

```python
len(filenames)
```

```python
# some of the lat-lon pairs above don't have fields (e.g. because they are over water)
wrong_filenames = []
for filename in filenames:
    if not os.path.isfile(filename):
        wrong_filenames.append(filename)
filenames = list(set(filenames)-set(wrong_filenames))
print(len(filenames))
```

```python
random.shuffle(filenames)

header = True
for filename in tqdm(filenames, smoothing=0):
    gdf = gpd.read_parquet(filename)
    gdf.to_csv('baseline_comparison_data/taylor_geospatial/taylor_geospatial_US_fields.csv', mode='a', index=False, header=header)
    header = False
```
